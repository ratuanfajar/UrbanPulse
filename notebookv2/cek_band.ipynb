```python
import os
import glob
import rasterio

# Jalur folder raw Jabar
folder_jabar = "../data/raw/jabar"

# Mencari seluruh file .tif di folder tersebut secara otomatis
daftar_file = glob.glob(os.path.join(folder_jabar, "*.tif"))

if not daftar_file:
    print(f"⚠️ Waduh, folder '{folder_jabar}' tidak ditemukan atau tidak berisi file .tif!")
    print("Silakan cek kembali apakah posisi folder 'data/raw/jabar' sudah benar dari arah notebookv2.")
else:
    # Ambil file pertama secara otomatis untuk dicek
    file_target = daftar_file[0]
    print(f"🔍 Berhasil menemukan file: {os.path.basename(file_target)}")
    print("-" * 50)
    
    with rasterio.open(file_target) as src:
        print(f"📊 Total Jumlah Band: {src.count}")
        print("📋 Nama-nama Band (Descriptions):")
        
        # Cek apakah band memiliki label nama text
        has_description = False
        for i in range(1, src.count + 1):
            name = src.descriptions[i-1]
            if name:
                print(f"   Band {i}: {name}")
                has_description = True
            else:
                print(f"   Band {i}: [Tidak ada nama teks]")
                
        if not has_description:
            print("\n💡 Catatan: GEE Anda mengekspor band tanpa teks description.")
            print("   Biasanya urutannya adalah urutan default satelit (misal Sentinel-2: B2, B3, B4, B8, dst).")
```
